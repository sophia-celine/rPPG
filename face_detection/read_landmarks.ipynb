```python
import h5py


landmarks_path = r"A:\dataset_raw\2355855_0\1_0_landmarks.h5"
with h5py.File(landmarks_path, "r") as landmark_data:
    # HDF5 datasets are read by name. Row i in each dataset corresponds
    # to zero-based frame i of the anonymized video.
    # -------------- File attributes ----------------------
    print(landmark_data.attrs.keys())
    print("Coordinate order:", landmark_data.attrs["coordinate_order"])
    print("Coordinate system:", landmark_data.attrs["coordinate_system"])
    print("FPS: ", landmark_data.attrs["fps"])
    print("Number of frames: ", landmark_data.attrs["frame_count"])
    print("Frame width: ", landmark_data.attrs["frame_width"])
    print("Frame height: ", landmark_data.attrs["frame_height"])
    print("Frame indexing: ", landmark_data.attrs["frame_indexing"])
    print("Number of landmarks: ", landmark_data.attrs["landmark_count"])
    print("Video filename: ", landmark_data.attrs["video_filename"])
    # -------------- Data content ----------------------
    print(landmark_data.keys())
    print("Face detected", landmark_data["face_detected"][:5])
    # face_detected has a boolean value for each frame. This value is True if the face was detected in the frame. This value is false
    # if no face was detected in the frame, in which case the landmarks of the last frame are used to build the face mask.
    print("Landmarks: ", landmark_data["landmarks"])
    # landmarks are x,y,z coordinates. Each point is a face landmark. Usually, there should be 478 landmarks for each frame.
    # The landmarks data vector has shape (n, 478, 3) where n is the total number of frames in the source video.
    print("Source frame: ", landmark_data["source_frame"])
    # source_frame has shape (n,) where n is the total number of frames in the source video. Each value is the index of the frame in 
    # the source video that was used to build the anonymized current frame. For example, usually if frame 2 has no face detected,
    # frame 1 is used to build the face mask, so the source frame for frame 2 is 1. If frame 2 has a face detected, then the source frame for frame 2 is 2.
    # -------------- Blur regions ----------------------
    print("Blur regions: ", landmark_data["blur_regions"].keys())
    # Blur regions are divided in 2: polygonal blur regions outside the face and face blur regions.
    # All videos have face blur regions, composed of right and left eyes and eyebrows, nose and mouth which are based on face landmarks. 
    # All videos also have a ring blur region based on landmarks of the border of the face, excluding a region linking a point in the right
    # area of the chin and the bottom right corner of the frame and a point in the left area of the chin and the bottom left corner of the frame.
    # Some videos have additional polygonal blur regions outside the face, which are manually annotated. For those, 'polygon_offsets' indicates
    # where the list of each polygon points starts and ends in the 'polygon_points' dataset. The 'polygon_points' dataset contains the x,y coordinates 
    # of all polygon points.

    print("Face blur: ", landmark_data["blur_regions"]["face_blur"].keys())
```
